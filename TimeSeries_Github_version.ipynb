```python
# Merge sentiments with locations in the dataframe on city
df_S  = pd.read_csv('/your_path/sentiments_file.csv', index_col=0, dtype='unicode')

loc = pd.read_csv('/your_path/locations_file.csv', index_col=0, dtype='unicode')

df_S = df_S.merge( loc, on='City' )   

# Fix the date in unified YearMonth format 
def get_year_month( row ):
    return( row.Date[:7])

def fix_date( row ):
    return( row.Date[:10])
    
df_S['YearMonth']  = df_S.apply( get_year_month, axis=1)
df_S['Date']  = df_S.apply( fix_date, axis=1)
```

```python
df_X = pd.DataFrame( df_S[ df_S[ 'Type' ].eq( 'X' ) ])
df_X.sample(10)
```

### Plot Time Series for the Data

```python
import plotly.express as px
import pandas as pd

# Plot sentiments grouped by YearMonth (use sum method)
fig = px.line(df_X.groupby('YearMonth').agg(
                Positive=('Sentiment', lambda x: (x == 'positive').sum()),
                Negative=('Sentiment', lambda x: (x == 'negative').sum()),
                Neutral=('Sentiment', lambda x: (x == 'neutral').sum())
).reset_index(), x='YearMonth', y=['Positive','Negative', 'Neutral'], title='X Corpus Time Series')

# Specify time ranges in the plot
fig.update_xaxes(
    rangeslider_visible=True,
    rangeselector=dict(
        buttons=list([
            dict(count=1, label="1m", step="month", stepmode="backward"),
            dict(count=6, label="6m", step="month", stepmode="backward"),
            dict(count=1, label="YTD", step="year", stepmode="todate"),
            dict(count=1, label="1y", step="year", stepmode="backward"),
            dict(step="all")
        ])
    )
)
fig.write_html("X_timeseries.html")
fig.show()              
```
