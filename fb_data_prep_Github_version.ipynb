### Facebook Data was obtained from the front end by copying and pasting threads of posts in selected pages and groups 

```python
import pandas as pd
import re
import os

# Create a folder for raw files of the Facebook pages and groups copied and pasted on documents and named it "data"
# Create a folder for the converted text files and named it "out"

path = "insert_your_folder_path_here"

# Create a for loop to iteratively open each file in "data" and save it in the "out" folder as text files

for myfile in os.listdir(path):
    if( myfile.endswith(".txt") ):
        #print( "processing ./data/" + myfile);
        f  = open("./data/" + myfile, "r" );
        fo = open("./out/" + myfile , "w" );

# Create an inside loop to separate posts by identifiers and isolate them
        for line in f.readlines():          
            if line.strip() not in ['\n','\r\n', ' ']:  
                if( '1y' in line or '2y' in line or '3y' in line or '4y' in line or 'Like' in line or '2w' in line ):
                    line = 'break_here'; 
                else:
                ##line = re.sub('^\· :$|\..*|https://.*$|y2|, of ,|View more comments|Write a comment|Write a reply…|: / :|…|[\d_]+|Follow|Like|Reply|See translation|Write a reply|\.\.\.|^[\t|\' \'].*$', ' ', line)        
                    ##line = re.sub('^\d+|^.*Share.*$|^View all \d+ replies$|^View 1 reply$|^\· \:$|https://.*$|, of ,|Edited|View more comments|Write a comment|Write a reply…|: / :|…|Follow|Like|Reply|Write a public reply…|See translation|Write a reply|\.\.\.|Most Relevant is selected, so some replies may have been filtered out.|Most relevant is selected, so some comments may have been filtered out.', ' ', line);
                    line = ' '.join( re.findall(r'[\u0600-\u06FF]+', line.strip()) )
                                        
                #if( len(line) != 0 and not line[0].isascii() ):
                if( len(line.strip()) != 0  ):
                    print( line.strip(), file=fo );
        print('break_here', file=fo);
```
