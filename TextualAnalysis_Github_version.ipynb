```python
# Create a matrix of sentiments by city

import matplotlib.pyplot as plt
from collections import Counter
import pprint as pp

# Build a matrix for sentiments by cities
titles = ['sentiment', 'city', 'country', 'region'] 
sentiment_type = ['positive', 'negative', 'neutral'] 

def build_matrix(sentiments):
    matrix = {}

    for row in sentiments.values.astype(str).tolist():
        if not matrix.get(row[0]):
            matrix[ row[0] ] = {}

        if not matrix[ row[0] ].get( row[1] ) :
            matrix[ row[0] ] [ row[1] ] = {}
            matrix[ row[0] ] [ row[1] ] = 1      
        else:
            matrix[ row[0] ] [ row[1] ] += 1      
            
    return(matrix)
```

```python
# Create a data frame using pandas to compare the occurrence of each sentiment by city for the X corpus

df = pd.DataFrame( build_matrix( df_X[['Sentiment','City']] ))

# Plot Sentiments by city
ax= df.plot.bar(color = ('#1E90FF','#228B22', '#FF8C00'), figsize=(20, 10))
ax.set_xlabel('Cities', fontsize=20)
ax.set_ylabel('Frequency', fontsize=20)
plt.xticks(fontsize=12)
plt.legend(fontsize = 20)
plt.savefig(FIGS_PATH+"/sentiments_by_city_X.png")
plt.show()

print(datetime.now() - t)
```

```python
# Import mle and tokenizer from CAMeL Tools (they were imported previously but we need them in this notebook to continue the analysis)
from camel_tools.disambig.mle import MLEDisambiguator
from camel_tools.tokenizers.morphological import MorphologicalTokenizer

# Load a pretrained disambiguator to use with a tokenizer
mle = MLEDisambiguator.pretrained('calima-msa-r13')

# By specifying `split=True`, the morphological tokens are output as seperate
# strings.
# We can output diacritized tokens by setting `diac=True`
tokenizer = MorphologicalTokenizer(mle, scheme='d3tok', split=True, diac=True)
```

```python
# Tokenize lines in each corpus
lines_X = tokenizer.tokenize( df_X['Text'].sample(10))
lines_fb = tokenizer.tokenize( df_fb['Text'].sample(10))

# Tokenize words in lines in each corpus
all_words_X = []
for line in lines_X:
    all_words_X.extend( line.split() )
    
all_words_fb = []
for line in lines_fb:
    all_words_fb.extend( line.split() )
```

```python
# Count the words in each corpus before stopwords
words_X  = sum(Counter(all_words_X).values())
words_fb = sum(Counter(all_words_fb).values())
```

```python
file_path='/your_path'

## Write words into files for each corpus
## We added this step in order to be able to run the below processing without repeating the prior steps
try:
    with open( file_path + 'all_words_X.txt', 'w') as f:
        for line in all_words_X:
            f.write(f"{line}\n")

    with open( file_path + 'all_words_fb.txt', 'w') as f:
        for line in all_words_fb:
            f.write(f"{line}\n")

except FileNotFoundError:
    print(f"Error: The file '{file_path}' was not found.")
    
```
